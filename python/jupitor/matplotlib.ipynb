```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2 * np.pi, 200)
y = np.sin(x)

fig, ax = plt.subplots()
ax.plot(x, y)
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 90, 90]
y = [10, 20, 9, -1]

# linestyle = -- - :
# marker = o * s
plt.plot(x, y, linestyle=':', marker='s')
plt.title("Panicking Rate")
plt.xlabel("month")
plt.ylabel("sale")

plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 90, 90]
y = [10, 20, 9, -1]
x_ = [10, 92, 9, 90]
y_ = [10, 20, 9, 100]

# linestyle = -- - :
# marker = o * s
plt.plot(x, y, linestyle=':', marker='s')
plt.plot(x, y_, linestyle=':', marker='s')
plt.plot(x_, y_, linestyle=':', marker='s')
plt.plot(x, y_, linestyle=':', marker='s')

plt.title("Panicking Rate")
plt.xlabel("month")
plt.ylabel("sale")

plt.show()
```

```python
import matplotlib.pyplot as plt

x_ = ['A', 'B', 'C']
y_ = [10, 20, 90]

plt.bar(x_, y_)
plt.show()
```

```python
import matplotlib.pyplot as plt

x_ = ['A', 'B', 'C']
y_ = [10, 20, 90]

plt.barh(x_, y_)
plt.show()
```

```python
import matplotlib.pyplot as plt

x_ = ['A', 'B', 'C']
y_ = [10, 20, 90]

plt.pie(y_, labels=x_, autopct='%1.1f%%')

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy

data = numpy.random.randn(100)
data_ = numpy.random.randn(100)

plt.hist(data, bins=90)
plt.hist(data_, bins=90)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy

data = numpy.random.randn(100)
data_ = numpy.random.randn(100)

plt.scatter(data, data_)
plt.scatter(data_, data)

plt.show()
```

```python
import matplotlib.pyplot as plt

temp = [31, 20, 30, 30, 23, 34, 23]
day = [1, 2, 3, 4, 5, 6, 7]

# linestyle = -- - :
# marker = o * s
plt.plot(day, temp, linestyle='-', marker='*')
plt.title("Panicking Rate")
plt.xlabel("month")
plt.ylabel("sale")
plt.grid('true')
plt.figure(figsize=(90,90))

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.subplot(1, 2, 1)
x = ['A', 'B', 'C']
y = [10, 20, 90]
plt.plot(x, y)

plt.subplot(1, 2, 2)
x_ = ['A', 'B', 'C']
y_ = [10, 20, 90]

plt.barh(x_, y_)
plt.show()
plt.savefig("salefig.png")
```
