```python
def firstN(n):
    i = 0
    while (i < n):
        yield i
        i += 1

print(list(firstN(10)))
```

```python
def add(a, b):
    return a + b
    
def sub(a, b):
    return a - b

def mul(a, b):
    return a * b

def getVal():
    a = int(input("a = "))
    b = int(input("b = "))
    return a, b

a, b = getVal()

print("add = ", add(a, b))

print("sub = ", sub(a, b))

print("mul = ", mul(a, b))
```
