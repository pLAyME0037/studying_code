```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

hours = np.array([1, 2, 3, 4, 5, 6]).reshape(-1, 1) 

scores = [50, 55, 65, 70, 80, 85]

model = LinearRegression()
model.fit(hours, scores)

print(f"Slope (Coefficient): {model.coef_[0]:.2f}")
print(f"Intercept:           {model.intercept_:.2f}")

test_hour = np.array([[10]])
predicted_score = model.predict(test_hour)
print(f"\nPredicted score for 10 hours: {predicted_score[0]:.2f}")

plt.scatter(hours, scores, color="blue", label="Data Points")
plt.plot(hours, model.predict(hours), color="red", label="Regression Line")
plt.xlabel("Hours")
plt.ylabel("Score")
plt.title("Hours vs Score (Linear Regression)")
plt.legend()
plt.grid(True)
plt.show()
```
